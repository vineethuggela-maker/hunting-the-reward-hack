```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
n_values = [1, 2, 4, 8, 16, 32, 64]
proxy, gold = [], []

for n in n_values:
    quality = np.random.normal(0.5, 0.2, (1000, n))
    reward = quality + np.random.normal(0, 0.3, (1000, n))
    best = np.argmax(reward, axis=1)

    proxy.append(np.mean(np.max(reward, axis=1)))
    gold.append(np.mean(quality[np.arange(1000), best]))

plt.plot(n_values, proxy, "r-o", label="Proxy Reward")
plt.plot(n_values, gold, "b-o", label="True Quality")
plt.xscale("log", base=2)
plt.xlabel("Number of Candidates")
plt.ylabel("Average Score")
plt.title("Hunting the Reward Hack")
plt.legend()
plt.grid()
plt.show()
```

Abstract

Reward models are widely used to guide language models toward outputs that humans prefer. However, a model that receives a high score from a learned reward function does not necessarily produce genuinely better answers. When optimization becomes too aggressive, the model may exploit weaknesses in the reward function instead of improving the underlying quality of its responses. This phenomenon is known as reward hacking or reward over-optimization.

This project investigates the relationship between proxy reward and genuine answer quality. The experiment varies the number of candidate responses considered during optimization and compares reward models trained with different quantities of preference data. Evaluation includes gold reward, proxy reward, response length, uncertainty intervals, and an independent judge.

The central research question is whether stronger optimization consistently improves answer quality or eventually produces outputs that score well under the proxy reward but perform poorly under the gold reward. The project is designed to provide a reproducible demonstration of reward hacking and to identify practical methods for detecting it.

1. Introduction

Reinforcement learning from human feedback and preference-based optimization are important approaches to aligning language models with human expectations. Instead of defining every desirable response through manually written rules, developers collect preferences between candidate answers and train a reward model to estimate which answers people prefer.

The resulting reward model provides a numerical objective for improving the language model. However, this approach creates a potential mismatch between the learned objective and the actual goal. The reward model is only an approximation of human preferences. It may learn correlations that are useful in the training dataset but unreliable for unfamiliar responses.

For example, a reward model might associate longer explanations, confident language, or excessive agreement with higher-quality answers. An optimized model could then produce unnecessarily long answers or flatter the reader, even when these behaviors reduce usefulness and factual quality.

This problem is important because successful optimization of a proxy objective can create a misleading impression of progress. An increase in predicted reward is not sufficient evidence that alignment has improved.

This project studies this issue through a controlled experiment. It uses a held-out gold reward to evaluate genuine quality independently of the proxy reward being optimized.

4. Background and Related Concepts
4.1 Reward models

A reward model estimates the quality or preference of an answer. In preference-based learning, training examples can contain two answers to the same prompt and a label indicating which answer is preferred.

A reward model learns to assign a higher score to preferred answers. Its predictions are useful only to the extent that its training data and learned representations capture the intended quality criteria.

4.2 Reward hacking

Reward hacking occurs when an optimized system obtains high reward by exploiting limitations in the objective rather than achieving the intended outcome.

In language generation, examples may include unnecessary verbosity, excessive flattery, repeated phrases that correlate with high scores, and confident claims unsupported by evidence.

A successful reward-hacking experiment should demonstrate a measurable difference between the proxy objective and an independently defined measure of quality.

4.3 Over-optimization

Increasing optimization strength can initially improve the selected response. As optimization becomes stronger, however, the system increasingly selects unusual responses that happen to receive high proxy scores.

The selected response may therefore have a high proxy reward but a lower gold reward. This creates an over-optimization curve in which genuine quality rises, reaches a peak, and then declines.

4.4 Preference data and generalization

A reward model trained on a larger and more representative dataset may generalize better than one trained on limited data. However, additional data does not guarantee that every weakness disappears. Model capacity, dataset quality, distribution shift, and the optimization procedure can all influence the outcome.

5.2 Gold reward

5.3 Proxy reward

The proxy reward is the objective used to select candidate answers. It can be a learned preference model trained on labeled comparisons.

For an initial simulation, a deliberately imperfect proxy can reward both answer quality and a misleading feature, such as flattery. This makes the reward mismatch measurable and allows the experiment to be tested before using a more computationally expensive model.

The simulation is a proof of concept, not evidence that a real language model has been aligned or trained successfully.

8. Discussion

The experiment illustrates an important distinction between optimizing a measurable objective and achieving the intended goal. A proxy reward model provides a convenient numerical signal, but it cannot automatically guarantee that the selected outputs satisfy the original quality criteria.

The results should be interpreted in the context of the experimental setup. A synthetic experiment can clearly demonstrate how reward mismatch creates over-optimization, but it does not establish the frequency or magnitude of reward hacking in deployed language models.

If additional preference data reduces the decline in gold reward, this would suggest that better reward estimation can help in the tested conditions. If the decline remains, it would demonstrate that improving reward-model training alone may not eliminate the problem.

Length controls are also important because a model may improve one metric by generating longer answers. Similarly, independent evaluation can reveal whether high proxy scores correspond to useful responses or merely to features that the proxy rewards.

9. Limitations

First, a synthetic reward function may simplify the complexity of human preferences. Results from a controlled simulation should not be generalized directly to large language models.

Second, the gold reward may fail to capture every relevant aspect of answer quality. An imperfect gold evaluator can itself introduce misleading conclusions.

Third, training multiple reward models and running repeated experiments may require additional computation. The report must disclose any reductions in training data, model size, or trial count.

Fourth, an independent automated judge may share biases with the proxy model. Blinded evaluation, a distinct rubric, and human checks can reduce but not completely eliminate this concern.

Finally, a peak observed over a limited range of optimization strengths may not be the global optimum. Additional experiments may be needed to establish the stability of the estimated peak.
